```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Example freshwater measurements
measurements = pd.DataFrame({
    "temperature_c": [18.2, 18.5, 19.1, 20.0, 20.4],
    "dissolved_oxygen_mg_l": [9.1, 8.9, 8.6, 8.2, 7.9],
    "pH": [7.8, 7.9, 8.0, 8.2, 8.3]
})

# Display the measurements
display(measurements)

# Plot temperature over the five observations
observations = np.arange(1, 6)

plt.figure(figsize=(8, 4))
plt.plot(
    observations,
    measurements["temperature_c"],
    marker="o"
)

plt.xlabel("Observation")
plt.ylabel("Temperature (°C)")
plt.title("Example Freshwater Temperature Measurements")
plt.grid(True)
plt.show()
```
